```python
# Importing necessary libraries
# Data based on https://www.youtube.com/watch?v=k8yIvVFCguA

from stoneforge.chat.forge_chat import ForgeChat
from stoneforge.petrophysics.shale_volume import gammarayindex
from pprint import pprint
import pandas as pd
import numpy as np
```

```python
chat = ForgeChat()
chat.register_decorated_tool(gammarayindex)

result = chat.call_tool(
    "gammarayindex",
    gr=np.array([64]),
    grmin=20,
    grmax=110
)

print(result)
```
